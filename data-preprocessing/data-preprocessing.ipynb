```python
import pandas as pd, matplotlib.pyplot as plt, numpy as np
```

```python
df = pd.read_csv('../data/hotel_booking_clean.csv')
df.head()
```

## Objectives

* Remove the negative value in ADR
* Merge weekend+ weekdays as total_stay
* Merge adults + children + babies as total_attendance
* Keep the selected features

### Remove Negative Value in ADR

```python
df[df['adr'] < 0]
```

```python
df = df[df['adr'] > 0].copy()
```

```python
df['adr'].min()
```

### Merge Weekend + WeekDays as total_stay

```python
df['total_stay'] = (df['stays_in_weekend_nights'] + df['stays_in_week_nights'])
```

```python
df[['total_stay', 'stays_in_weekend_nights','stays_in_week_nights']].head()
```

```python
df.drop(columns=['stays_in_weekend_nights','stays_in_week_nights'], inplace=True)
```

### Merge babies + adults + children as total_attendance

```python
df['total_attendance'] = (df['babies'] + df['adults'] + df['children']).astype(int)
```

```python
df[['total_attendance', 'babies','adults','children']].head()
```

```python
df.drop(columns=['babies','adults','children'], inplace=True)
```

### Keep the selected features

```python
selected_features = [
    "is_canceled",
    "lead_time",
    "arrival_date_week_number",
    "arrival_date_day_of_month",
    "total_stay",
    "total_attendance",
    "meal",
    "country",
    "market_segment",
    "reserved_room_type",
    "assigned_room_type",
    "deposit_type",
    "adr",
    "total_of_special_requests"
]

df = df[selected_features]
```
